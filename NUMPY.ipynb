**List vs Numpy - Time Taken**

```python
from time import process_time
```

Time taken by a list

```python
List=[i for i in range(10000)]

start_time=process_time()

List=[i+5 for i in List]

end_time=process_time()

print(end_time-start_time)
```

Time Taken by Numpy

```python
np_array=np.array([i for i in range(10000)])

start_time=process_time()

np_array+=5 #np_array=np_array+5

end_time=process_time()

print(end_time-start_time)
```

**Initial Placeholders in numpy arrays**

```python
#Creates a numpy array of zeros
x=np.zeros((4,5))
print(x)
```

```python
#Creates a numpy array of ones
y=np.ones((3,3))
print(y)
```

```python
#Creates a numpy array of a particular value
z=np.full((2,3),5)
print(z)
```

```python
#Create an Identity matrix
s=np.eye(3)
print(s)
```

```python
#Create a numpy array with random values
p=np.random.random((3,4))
print(p)
```

```python
#Create a numpy array with random integer values within a specific range
t=np.random.randint(10,100,(3,5))
print(t)
```

```python
#Create a numpy array of evenly spaced value --> specifying the number of values required
r=np.linspace(10,30,6)  #5 --> represents that i want 5 values between 10 to 30 evenly
print(r)
```

```python
#Create a numpy array of evenly spaced value --> specifying the step
a=np.arange(10,30,4)  #5--> represents the steps
print(a)
```

```python
#Convert a list to a numpy array
List2=[10,20,20,20,50]

np_array=np.asarray(List2)
print(np_array)
type(np_array)
```

Binary operations

```python
a=np.random.randint(0,10,(3,3))
b=np.random.randint(10,20,(3,3))
print(a)
print(b)
print(np.add(a,b))
print(np.subtract(a,b))
print(np.multiply(a,b))
print(np.divide(a,b))
```

Aggregation function

```python
n=np.random.randint(0,20,5)
print(n)
print(n.max())
print(n.min())
print(n.sum())
print(n.mean())
print(n.std())
```

Array Manipulation

```python
array=np.random.randint(0,10,(2,3))
print(array)
array.shape
```

```python
#Transpose
trans=np.transpose(array)
print(trans)
trans.shape
```

```python
array=np.random.randint(0,10,(2,3))
print(array)
array.shape
```
