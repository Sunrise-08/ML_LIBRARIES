**Pandas Library – Google Colab Notes**


Pandas is an open-source Python library used for data manipulation, analysis, and cleaning. It is built on top of NumPy and works directly in Google Colab.

**Key Features:**
1. Provides two data structures: Series (1D) and DataFrame (2D).
2. Handles missing and duplicate values.
3. Supports CSV and Excel files.
4. Helps in data filtering, sorting, and analysis.

Pandas Library: Useful for Data Procesing & Analysis


Pandas Data Frame: Panda Data Frame is two-dimensional tabular data structure with labeled axes(rows and columns).

Manipulating a DataFrame


```python
#adding a column to a DataFrame
california_df['Price']=california_dataset.target
```

```python
california_df.head()
```

```python
#removing a particular row
california_df.drop(index=0,axis=0)  #if axis=0, it will remove the row but if it is equal to 1 it will remove column
```

```python
#removing a particular column
california_df.drop(columns="HouseAge",axis=1)
```

```python
#locating a row using the index value
california_df.iloc[3]
```

```python
#locating a particular column
print(california_df.iloc[:,0])   #first column of the dataframe
print(california_df.iloc[:,1])   #second column of the dataframe
print(california_df.iloc[:,2])   #third column of the dataframe
print(california_df.iloc[:,-1])   #last column of the dataframe
```
